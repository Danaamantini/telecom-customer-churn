# 02 — Business analysis

SQL-driven KPIs, churn segments, and observed recurring revenue associated with churn. Associations are not interpreted as causal effects.

```python
from pathlib import Path
import duckdb
import matplotlib.pyplot as plt
import seaborn as sns

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
DB = ROOT / 'data' / 'processed' / 'churn.duckdb'
FIGURES = ROOT / 'reports' / 'figures'
FIGURES.mkdir(parents=True, exist_ok=True)
connection = duckdb.connect(str(DB), read_only=True)
kpis = connection.sql('SELECT * FROM v_kpis').df()
kpis
```

```python
segments = connection.sql("SELECT * FROM v_churn_by_segment").df()
contracts = segments[segments['segment'] == 'contract'].sort_values('churn_rate', ascending=False)
internet = segments[segments['segment'] == 'internet_type'].sort_values('churn_rate', ascending=False)
display(contracts)
display(internet)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))
sns.barplot(data=contracts, x='churn_rate', y='segment_value', ax=axes[0], color='#D64550')
axes[0].set(title='Churn rate by contract', xlabel='Churn rate', ylabel='')
sns.barplot(data=internet, x='churn_rate', y='segment_value', ax=axes[1], color='#4472C4')
axes[1].set(title='Churn rate by internet type', xlabel='Churn rate', ylabel='')
fig.tight_layout()
fig.savefig(FIGURES / 'business_segments.png', dpi=150, bbox_inches='tight')
plt.show()
connection.close()
```
